**Operators**

```python
# Arithmetric Operators
print(5+6) #addition
print(5-6) #substaction
print(5*6) #multiplication
print(5/2) #division
print(5//2) #integer division
print(5%2) #modulas-->reminder
print(5**2) #power off-->5 to the power 2
```

```python
#Reational Operators
print(4<5)
print(4>=4)
print(4<=4)
print(4==4)
print(4!=4)
```

```python
# Logical Operators
print(1 and 0) #0
print(1 or 0) #1
print(not 1) #0-->False
```

```python
#Bitwise Operators
#bitwise and
print(2&3) #2 --> 2->010, 3->110, dono mai & lagaoge then we get 010 i.e, 2

# biwise or
print(2|3) #3 -->2->010, 3->110, dono mai or lagaoge then we get 110 i.e, 3

# bitwise xor
print(2^3) #1 -->2->010, 3->110, dono mai xor lagaoge then we get 100 i.e, 1
# xor gives true when input are different.

# bitwise not
print(~2)

# leftsift
print(4>>2)

# rightsift
print(5<<2)
```
